```python
import matplotlib.pyplot as plt
import numpy as np

# Set seed for reproducible random noise
np.random.seed(42)

# 1. Parameters & Data Generation
n = 50
true_slope = 2
true_intercept = 5
noise_std = 1

x = np.linspace(0, 10, n)
y_raw = true_slope * x + true_intercept + np.random.normal(loc=0.0, scale=noise_std, size=n)

# Convert x and y into 2D matrices/vectors matching mathematical notation
X = x[:, None]              # Design matrix X, shape: (50, 1)
y = y_raw[:, None]          # Target vector y, shape: (50, 1)

# Adjusted target vector for fixed intercept model: y_adj = y - b
y_adj = y - true_intercept

# 2. Linear Algebra Approach (Normal Equation)
# Formula: w = (X^T * X)^(-1) * X^T * y_adj
w_matrix = (np.linalg.inv(X.T @ X) @ X.T @ y_adj).item()

# Check: Optimal Slope w
print(f"Matrix Inverse Slope:  w = {w_matrix:.15f}")
```
