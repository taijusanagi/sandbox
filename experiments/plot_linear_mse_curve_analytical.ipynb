```python
import matplotlib.pyplot as plt
import numpy as np

# Set seed for reproducible random noise
np.random.seed(42)

# 1. Parameters
n = 50
slope = 2
intercept = 5
noise_std = 1

# 2. Data Generation
x = np.linspace(0, 10, n)
y = slope * x + intercept + np.random.normal(loc=0.0, scale=noise_std, size=n)

# 3. Candidate Slopes Range
slopes = np.linspace(-1, 5, 200)

# --- ANALYTICAL PARABOLA COEFFICIENTS ---
# Error for point i:  (w * x_i + b - y_i)
# Let c_i = (b - y_i)
# Squared error_i:    (w * x_i + c_i)^2 = (x_i^2) * w^2 + (2 * x_i * c_i) * w + c_i^2
# Taking the mean:    f(w) = A * w^2 + B * w + C
b_minus_y = intercept - y

A = np.mean(x**2)
B = 2 * np.mean(x * b_minus_y)
C = np.mean(b_minus_y**2)

# Analytical MSE function
def f(w):    
    return A * (w**2) + B * w + C

# Find the minimum MSE point
# f'(w) = 2A*w + B = 0  =>  w = -B / (2A)
optimal_w = -B / (2 * A)
min_mse = f(optimal_w)

# 4. Visualization
plt.figure(figsize=(8, 5))
plt.plot(slopes, f(slopes), color='blue', label='MSE Loss Curve')
plt.scatter([optimal_w], [min_mse], color='red', s=20, zorder=10, label=f'Minimum MSE ({optimal_w:.2f})')

plt.xlabel('Slope (w)')
plt.ylabel('Mean Squared Error (MSE)')
plt.title('MSE Loss Curve')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
