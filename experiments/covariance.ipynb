```python
import numpy as np

# Set seed for reproducible random noise
np.random.seed(42)

# Data generation
n = 50
slope = 2
intercept = 5
noise_std = 1

x = np.linspace(0, 10, n)
y = slope * x + intercept + np.random.normal(loc=0.0, scale=noise_std, size=n)

# Pre-compute means
mean_x = np.mean(x)
mean_y = np.mean(y)

# 1. Benchmark (NumPy default divides by n-1; bias=True forces division by n)
# Note: np.cov returns a 2x2 matrix [[Var(x), Cov(x,y)], [Cov(y,x), Var(y)]]
cov_matrix = np.cov(x, y, bias=True)
print("NumPy cov:", cov_matrix[0, 1])

# 2. Manual calculation using a for-loop
sum_prod_diff = 0
for x_val, y_val in zip(x, y):
    sum_prod_diff += (x_val - mean_x) * (y_val - mean_y)
print("For-loop cov:", sum_prod_diff / n)

# 3. Vectorized calculation using a dot product
x_centered = x - mean_x
y_centered = y - mean_y
print("Dot-product cov:", (x_centered @ y_centered) / n)
```
