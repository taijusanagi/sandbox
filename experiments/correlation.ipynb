```python
import numpy as np

# Set seed for reproducible random noise
np.random.seed(42)

# Data generation
n = 50
slope = 2
intercept = 5
noise_std = 1

x = np.linspace(0, 10, n)
y = slope * x + intercept + np.random.normal(loc=0.0, scale=noise_std, size=n)

# Pre-compute means
mean_x = np.mean(x)
mean_y = np.mean(y)

# 1. Benchmark (NumPy corrcoef returns a 2x2 matrix [[Corr(x,x), Corr(x,y)], [Corr(y,x), Corr(y,y)]])
corr_matrix = np.corrcoef(x, y)
print("NumPy corr:", corr_matrix[0, 1])

# 2. Manual calculation using a for-loop
sum_prod_diff = 0
sum_sq_diff_x = 0
sum_sq_diff_y = 0

for x_val, y_val in zip(x, y):
    dx = x_val - mean_x
    dy = y_val - mean_y
    sum_prod_diff += dx * dy
    sum_sq_diff_x += dx ** 2
    sum_sq_diff_y += dy ** 2

for_loop_corr = sum_prod_diff / np.sqrt(sum_sq_diff_x * sum_sq_diff_y)
print("For-loop corr:", for_loop_corr)

# 3. Vectorized calculation using a dot product
x_centered = x - mean_x
y_centered = y - mean_y

dot_corr = (x_centered @ y_centered) / np.sqrt(
    (x_centered @ x_centered) * (y_centered @ y_centered)
)
print("Dot-product corr:", dot_corr)
```
