```python
import matplotlib.pyplot as plt
import numpy as np

# Set seed for reproducible random noise
np.random.seed(42)

# 1. Parameters
n = 50
slope = 2
intercept = 5
noise_std = 1

# 2. Data Generation
x = np.linspace(0, 10, n)
y = slope * x + intercept + np.random.normal(loc=0.0, scale=noise_std, size=n)

# 3. Candidate Slopes & MSE Calculation
slopes = np.linspace(-1, 5, 200)

predictions = x[:, None] * slopes[None, :] + intercept
mse = np.mean((predictions - y[:, None]) ** 2, axis=0)

# Find the minimum MSE point
best_idx = np.argmin(mse)
best_slope = slopes[best_idx]
min_mse = mse[best_idx]

# 4. Visualization
plt.figure(figsize=(8, 5))
plt.plot(slopes, mse, color='blue', label='MSE Loss Curve')
plt.scatter([best_slope], [min_mse], color='red', s=20, zorder=10, label=f'Minimum MSE ({best_slope:.2f})')

plt.xlabel('Slope (w)')
plt.ylabel('Mean Squared Error (MSE)')
plt.title('MSE Loss Curve')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()
```
