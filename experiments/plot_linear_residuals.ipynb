```python
import matplotlib.pyplot as plt
import numpy as np

# Set seed for reproducible random noise
np.random.seed(42)

# 1. Parameters
slope = 2
intercept = 5
noise_std = 5  # Standard deviation of the error (noise)

# 2. Data Generation
n = 50
x_input = np.arange(n)

# Theoretical baseline line: y = slope * x + intercept
y_true = slope * x_input + intercept

# Measurement noise / residual errors
residuals = np.random.normal(loc=0, scale=noise_std, size=n)

# Actual observed data
y_obs = y_true + residuals

# 3. Visualization
plt.plot(x_input, y_true, label=f'True Line (y = {slope}x + {intercept})', color='blue')
plt.scatter(x_input, y_obs, label='Observed Data', color='red', zorder=3)
plt.vlines(x_input, ymin=y_true, ymax=y_obs, colors='gray', linestyles='dotted', label='Residuals')

plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
