```python
import numpy as np

# Set seed for reproducible random noise
np.random.seed(42)

# Data generation
n = 50
x = np.linspace(0, 10, n)

# 1. Benchmark (NumPy default: population variance, divides by n)
print("NumPy var:", np.var(x))

# 2. Manual calculation using a for-loop
sum_sq_diff = 0
for val in x:
    sum_sq_diff += (val - x.mean()) ** 2
print("For-loop var:", sum_sq_diff / n)

# 3. Vectorized calculation using a dot product
x_centered = x - x.mean()
print("Dot-product var:", (x_centered @ x_centered) / n)
```
